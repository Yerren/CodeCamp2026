# Naive prompt
Just ask the model, with no information about valid applications or output format.

```python
import json
import time

from dotenv import load_dotenv
from google import genai
from google.genai import types
from IPython.display import JSON, Markdown, display

load_dotenv()  # loads GEMINI_API_KEY from .env
client = genai.Client()
```

```python
transcript = "Hey Google, navigate to Ara."
```

## Input

```python
message = f"""For the following transcript, determine the target action the user wants to take,
the target application, and any additional parameters.\n transcript: {transcript}"""
print(message)
```

## Model call

```python
start = time.perf_counter()
interaction = client.interactions.create(
    model="gemini-3.5-flash-lite",
    input=message,
)
print(f"Client call took {time.perf_counter() - start:.2f}s")
```

## Output

```python
display(Markdown(interaction.output_text))
```
